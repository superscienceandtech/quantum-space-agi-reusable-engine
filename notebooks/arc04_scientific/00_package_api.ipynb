```python

import sys
from pathlib import Path
repo_root = next(
    (
        parent
        for parent in (Path.cwd(), *Path.cwd().parents)
        if (parent / "src" / "quantum_space_agi").is_dir()
    ),
    None,
)
if repo_root is not None:
    sys.path.insert(0, str(repo_root / "src"))
import numpy as np
from quantum_space_agi.scientific import (
    CoordinateFrame,
    ScientificError,
    Unit,
    active_rotation_matrix_z,
    create_transform,
    euler_rotation_matrix,
    get_unit,
    transform_point,
)
assert CoordinateFrame.__name__ == "CoordinateFrame"
assert ScientificError.__name__ == "ScientificError"
assert Unit.__name__ == "Unit"
assert callable(active_rotation_matrix_z)
assert callable(euler_rotation_matrix)
assert get_unit("m").symbol == "m"
frame = CoordinateFrame("A")
rotation = active_rotation_matrix_z(np.pi / 2.0)
transform = create_transform(frame, frame, rotation=rotation, translation=np.array([1.0, 2.0, 0.0]))
point = np.array([1.0, 0.0, 0.0], dtype=np.float64)
result = transform_point(transform, point)
assert np.allclose(result, np.array([1.0, 3.0, 0.0], dtype=np.float64))
```
