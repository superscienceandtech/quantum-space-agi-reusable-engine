```python
from quantum_space_agi.scientific.errors import ScientificError, ScientificErrorCode
try:
    raise ScientificError(ScientificErrorCode.INVALID_STATE, 'Raised intentionally for notebook demonstration.', field='demo', expected='valid input', received='invalid')
except ScientificError as exc:
    print('code:', exc.code.value if hasattr(exc.code, 'value') else exc.code)
    print('message:', exc.message)
    print('field:', exc.field)
    print('metadata:', {'expected': exc.expected, 'received': exc.received})
```
